```python
import numpy as np
import xarray as xr
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
filename = Path(r"C:/Users/karoa/MOHID_internship/MOHID_model_workflow/data/preprocessing/26428_55105_3463_SEA_LEVEL_20230113110137_20230223110137.csv")
output_folder = filename.parent
# --- 1. Read the input file ---
df = pd.read_csv(
    filename,
    sep="\t",          # Tab-separated
    skiprows=1,        # Skip the metadata line
    na_values=-9999.9, # Convert the null value to NaN
    encoding="utf-8"
)

df['datetime'] = pd.to_datetime(df['Fecha (GMT)']) 
```

```python
# --- 2. Convert cm to m
df['water_level'] = df['Nivel (cm)'] / 100

# --- 3. Build the SECONDS column relative to the first timestamp ---
t0 = df['datetime'].iloc[0]
df['seconds'] = (df['datetime'] - t0).dt.total_seconds().astype(int)

# --- 4. Write the output file ---
# build fileout name
start = df['datetime'].iloc[0]
end = df['datetime'].iloc[-1]

fileout = output_folder / f"WATER_LEVEL_{start.year}_{start.month}_{start.day}_{end.year}_{end.month}_{end.day}.dat"

initial = t0.strftime('%Y. %m. %d. %H. %M. %S.').replace(' 0', ' ')  
# Safer: build the header explicitly from the datetime parts
initial = f"{t0.year}. {t0.month:>2}. {t0.day:>2}. {t0.hour}. {t0.minute}. {t0.second}."
```

```python
# save as csv
gauge = pd.DataFrame({
    'date': df.datetime,
    'water_level': df.water_level
})
gauge.to_csv(r'C:/Users/karoa/MOHID_internship/MOHID_model_workflow/data/preprocessing/WATER_LEVEL_2023_1_13_2023_2_23.csv', index=False)
```
